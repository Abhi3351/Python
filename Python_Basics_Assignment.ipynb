```python
#Question 2 : Explain the difference between List and Tuple in Python.
```

```python
#Below are the main difference between List and Tuple.
```

```python
#1. List is Mutabile
#A List is mutable, which means its elements can be added, removed, or modified after creation.
my_list = [1, 2, 3]
my_list.append(4)

print(my_list)
```

```python
##1. Tuple is Immutable
#A Tuple is immutable, which means its elements cannot be changed after creation.
my_tuple = (1, 2, 3)
my_tuple[0] = 10  # Error
```

```python
##2. List's Syntax is square brackets[]
L_students = ["Aman", "Riya", "Rahul"]

##2. Whether Tuples are created using parentheses()
T_students = ("Aman", "Riya", "Rahul")
```

```python
print(L_students)
print(T_students)
```

```python
##3. Performance
#Lists are slightly slower because they allow modifications.
#Tuples are faster because their values cannot be changed once created.
```

```python
##4. Memory Usage
#Lists consume more memory due to their mutable nature.
#Tuples consume less memory and are therefore more efficient for storing fixed data.
```

```python
##5. Use Case
#Use a List when data needs to be modified frequently.
#Use a Tuple when data should remain constant and secure throughout the program.
```

```python
#Question 3 : What is a function in Python? Why are functions useful?
```

```python
#A function in Python is a block of organized and reusable code that performs a specific task. 
#Functions help break a large program into smaller, manageable parts. A function is defined using 
#the def keyword and can be called whenever needed in the program.
```

```python
#Example of a function in Python
def abc():
    print("Hello, Welcome to Python buddy!")

abc()
```

```python
#Functions are useful...
```

```python
##1. Code Reusability
#Functions allow the same code to be used multiple times without rewriting it.

def add(a, b):
    return a + b

print(add(5, 3))
print(add(10, 7))
```

```python
##2. Reduces Code Duplication
#Instead of writing the same code repeatedly, a function can be called whenever required.
```

```python
##3. Improves Readability
#Functions make programs easier to read and understand by dividing them into smaller logical sections.
```

```python
##4. Easier Debugging and Maintenance
#If an error occurs, it can be fixed in one place inside the function instead of updating multiple
#sections of code.
```

```python
##5. Better Program Organization
#Functions help organize code efficiently, making large programs more structured and manageable.
```
