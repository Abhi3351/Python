```python
#Q4. Indexing using .loc
#Using .loc, perform the following:
#Display the row with index 10.
#Display the Sales value of the row with index 20
```

```python
#Q7. Filtering Data
#Filter the dataset to display:
#1. Rows where Region = "East".
#2. Rows where Sales > 1500.
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/retail_dataset .csv")

# 1. Rows where Region = 'East'
print("Rows where Region = 'East':")
print(df[df['Region'] == 'East'])

# 2. Rows where Price > 1500
print("\nRows where Price > 1500:")
print(df[df['Price'] > 1500])
```

```python
#Q8. Multiple Condition Filtering
#Display rows where:
#Region is West
#Sales is greater than 1200
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/retail_dataset .csv")

# Filter rows where Region is West and Price > 1200
filtered_data = df[(df['Region'] == 'West') & (df['Price'] > 1200)]

# Display result
print(filtered_data)
```

```python
#Q9. Descriptive Statistics
#Generate summary statistics using:
#describe()
#From the result identify:
#Mean sales value
#Maximum quantity sold
#Minimum sales value
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/retail_dataset .csv")

# Generate summary statistics
summary = df.describe()

print("Summary Statistics:")
print(summary)

# Mean Price (Sales)
print("\nMean Price:")
print(df['Price'].mean())

# Maximum Quantity Sold
print("\nMaximum Quantity Sold:")
print(df['Quantity'].max())

# Minimum Price (Sales)
print("\nMinimum Price:")
print(df['Price'].min())
```

```python
#Q10. Aggregation Analysis
#Using the Sales column, calculate:
#Total sales
#Average sales
#Maximum sales
#Minimum sales
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/retail_dataset .csv")

# Total Sales (Price)
total_sales = df['Price'].sum()

# Average Sales (Price)
average_sales = df['Price'].mean()

# Maximum Sales (Price)
max_sales = df['Price'].max()

# Minimum Sales (Price)
min_sales = df['Price'].min()

# Display results
print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales:", max_sales)
print("Minimum Sales:", min_sales)
```
