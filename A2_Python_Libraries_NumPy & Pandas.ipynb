```python
#Q1. Load the given retail sales dataset into a Pandas DataFrame and display the first 5 rows 
#    along with the dataset information to understand its structure.
```

```python
# Load the dataset
df = pd.read_csv("retail_dataset .csv")

# Display first 5 rows
print("First 5 Rows of the Dataset:")
print(df.head())

# Display dataset information
print("\nDataset Information:")
print(df.info())
```

```python
#Q2. Inspect the dataset and identify missing values in each column. Display the total number of missing 
# values present in the dataset.
```

```python
# Missing values in each column
print("Missing Values in Each Column:")
print(df.isnull().sum())

# Total missing values in the dataset
total_missing = df.isnull().sum().sum()

print("\nTotal Missing Values in Dataset:")
print(total_missing)
```

```python
#Q3. Replace missing values in the Quantity and Price columns with their respective mean values using Pandas.
```

```python
# Calculate mean values
quantity_mean = df["Quantity"].mean()
price_mean = df["Price"].mean()

# Replace missing values with mean
df["Quantity"].fillna(quantity_mean, inplace=True)
df["Price"].fillna(price_mean, inplace=True)

# Verify missing values are replaced
print("Missing Values After Replacement:")
print(df[["Quantity", "Price"]].isnull().sum())
```

```python
#Q4. Remove all rows where the Product Category or Region column contains missing values.
```

```python
# Remove rows with missing values in Product Category or Region
df = df.dropna(subset=["Product Category", "Region"])

# Verify missing values
print("Missing Values After Removal:")
print(df[["Product Category", "Region"]].isnull().sum())

# Check updated dataset size
print("\nNumber of rows after removal:", len(df))
```

```python
#Q5. Create a new column called Revenue using NumPy, where:
#Revenue = Quantity × Price
#Display the updated dataset with the new column.
```

```python
import numpy as np

# Create Revenue column
df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

# Display updated dataset
print("Dataset with Revenue Column:")
print(df.head())
```

```python
#Q6. Using NumPy, calculate the total revenue generated from all transactions in the dataset.
```

```python
import numpy as np

# Calculate total revenue
total_revenue = np.sum(df["Revenue"])

# Display total revenue
print("Total Revenue Generated:", total_revenue)
```

```python
#Q7. Group the dataset by Product Category and calculate the total revenue generated by each category.
```

```python
# Group by Product Category and calculate total revenue
category_revenue = df.groupby("Product Category")["Revenue"].sum()

# Display result
print("Total Revenue by Product Category:")
print(category_revenue)
```

```python
#Q8. Based on the revenue calculated for each product category, identify the top 3 categories with the 
#highest revenue and the bottom 3 categories with the lowest revenue.
```

```python
# Calculate total revenue by category
category_revenue = df.groupby("Product Category")["Revenue"].sum()

# Top 3 categories
top_3 = category_revenue.sort_values(ascending=False).head(3)

# Bottom 3 categories
bottom_3 = category_revenue.sort_values(ascending=True).head(3)

print("Top 3 Categories by Revenue:")
print(top_3)

print("\nBottom 3 Categories by Revenue:")
print(bottom_3)
```

```python
#Q9. Group the dataset by Region and calculate the total revenue generated in each region. Identify the 
#region with the highest revenue and the region with the lowest revenue.
```

```python
# Calculate total revenue by region
region_revenue = df.groupby("Region")["Revenue"].sum()

print("Revenue by Region:")
print(region_revenue)

# Region with highest revenue
highest_region = region_revenue.idxmax()
highest_revenue = region_revenue.max()

print("\nRegion with Highest Revenue:")
print(highest_region, "-", highest_revenue)

# Region with lowest revenue
lowest_region = region_revenue.idxmin()
lowest_revenue = region_revenue.min()

print("\nRegion with Lowest Revenue:")
print(lowest_region, "-", lowest_revenue)
```

```python
#Q10. Extract the month from the Date column using Pandas and group the dataset by month to calculate the 
#total monthly revenue. Also, calculate the mean, median, and standard deviation of the Revenue column 
#using NumPy.
```

```python

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Extract month name
df["Month"] = df["Date"].dt.month_name()

# Calculate monthly revenue
monthly_revenue = df.groupby("Month")["Revenue"].sum()

print("Monthly Revenue:")
print(monthly_revenue)

# Revenue statistics
mean_revenue = np.mean(df["Revenue"])
median_revenue = np.median(df["Revenue"])
std_revenue = np.std(df["Revenue"])

print("\nRevenue Statistics:")
print("Mean Revenue:", mean_revenue)
print("Median Revenue:", median_revenue)
print("Standard Deviation:", std_revenue)
```
